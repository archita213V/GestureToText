```python
X_train = train_df.iloc[:, :-1]
y_train = train_df.iloc[:, -1]
```

```python
X_test = test_df.iloc[:, :-1]
y_test = test_df.iloc[:, -1]
```

```python
X_train.shape
```

```python
y_train.shape
```

```python
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
# Encode training labels
y_train_encoded = label_encoder.fit_transform(y_train)
# Encode testing labels
y_test_encoded = label_encoder.transform(y_test)
print(y_train_encoded[:10])
```

```python
print(label_encoder.classes_)
print(label_encoder.transform(label_encoder.classes_))
```

```python
import pandas as pd
import joblib
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
lr_model = LogisticRegression(max_iter=1000)
lr_model.fit(X_train, y_train)
y_pred = lr_model.predict(X_test)
lr_accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", round(lr_accuracy * 100, 2), "%")
joblib.dump(lr_model, "../saved_models/logistic_regression.pkl")
print("Model Saved Successfully!")
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train, y_train)
y_pred2 = knn_model.predict(X_test)
knn_accuracy = accuracy_score(y_test, y_pred2)
print("Accuracy:", round(knn_accuracy * 100, 2), "%")
joblib.dump(knn_model, "../saved_models/knn.pkl")
print("KNN Model Saved Successfully!")
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)
y_pred = dt_model.predict(X_test)
dt_accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", round(dt_accuracy * 100, 2), "%")
joblib.dump(dt_model, "../saved_models/decision_tree.pkl")
print("Decision Tree Model Saved Successfully!")
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

```python
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)
rf_model.fit(X_train, y_train)
y_pred = rf_model.predict(X_test)
rf_accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", round(rf_accuracy * 100, 2), "%")
joblib.dump(rf_model, "../saved_models/random_forest.pkl")
print("Random Forest Model Saved Successfully!")
```

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

```python
svm_model = SVC()
svm_model.fit(X_train, y_train)
y_pred = svm_model.predict(X_test)
svm_accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", round(svm_accuracy * 100, 2), "%")
joblib.dump(svm_model, "../saved_models/svm.pkl")
print("SVM Model Saved Successfully!")
```

```python
import pandas as pd
results = {
    "Model": [
        "Logistic Regression",
        "KNN",
        "Decision Tree",
        "Random Forest",
        "SVM"
    ],
    "Accuracy": [
        lr_accuracy,
        knn_accuracy,
        dt_accuracy,
        rf_accuracy,
        svm_accuracy
    ]
}
results_df = pd.DataFrame(results)
results_df["Accuracy (%)"] = results_df["Accuracy"] * 100
results_df = results_df.sort_values(
    by="Accuracy (%)",
    ascending=False
)
results_df.reset_index(drop=True, inplace=True)
results_df
```
